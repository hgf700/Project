```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from ast import literal_eval
import ast

csv_path = r'C:\Users\USER098\Documents\GitHub\Project\RecommendationsML\Data\tmdb_5000\tmdb_5000_credits.csv'
pd.set_option('display.max_colwidth', None)
df = pd.read_csv(csv_path)

df["cast"] = df["cast"].apply(ast.literal_eval)
df["crew"] = df["crew"].apply(ast.literal_eval)

df["cast_names"] = df["cast"].apply(
    lambda x: [actor["name"].replace(" ", " ") for actor in x[:3]]  
)

df["cast_id"] = df["cast"].apply(
    lambda x: " ".join(str(actor["cast_id"]) for actor in x[:3])
)

df["cast_id"] = df["cast_id"].apply(lambda x: x.split())

df["order"] = df["cast"].apply(
    lambda x: " ".join(str(actor["order"]) for actor in x[:3])
)

df["order"] = df["order"].apply(lambda x: x.split())

def get_director(x):
    for i in x:
        if i['job'] == 'Director':
            return i['name']
    return np.nan
    
df['director'] = df['crew'].apply(get_director)

df.drop(columns=["crew","cast"], inplace=True)

df.to_csv('Data/results/tmdb_credits_edited.csv', index=False)
df.head()
```

```python
import ast

def combine_features(row):
    director = row['director'].replace(" ", "") if isinstance(row['director'], str) else ""
    overview = row['overview'] if isinstance(row['overview'], str) else ""
    
    try:
        genres = ast.literal_eval(row['genres'])
    except:
        genres = []
    
    try:
        production_companies = [name.replace(" ", "") for name in ast.literal_eval(row['production_companies'])]
    except:
        production_companies = []
    
    try:
        cast_names = [name.replace(" ", "") for name in ast.literal_eval(row['cast_names'])]
    except:
        cast_names = []

    try:
        year = str(pd.to_datetime(row['release_date']).year)
    except:
        year = ""
    
    return " ".join(genres) + " " + director + " ".join(cast_names) + " ".join(production_companies) + " " + overview + " " + year

df['tags'] = df.apply(combine_features, axis=1)

df.to_csv('Data/results/final.csv', index=False)

df.head(1)
```

```python
# from sklearn.feature_extraction.text import CountVectorizer
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

cv = TfidfVectorizer(
    max_features=20000,
    stop_words='english',
    ngram_range=(1,2)
)

# cv = CountVectorizer(max_features=5000, stop_words='english')

vectors = cv.fit_transform(df['tags']).toarray()

similarity = cosine_similarity(vectors)

def recommend(movie_title):
    movie_index = df[df['title'] == movie_title].index[0]
    distances = list(enumerate(similarity[movie_index]))
    distances = sorted(distances, key=lambda x: x[1], reverse=True)[1:6]  # top 5 podobnych
    for i in distances:
        print(df.iloc[i[0]]['title'])
```
